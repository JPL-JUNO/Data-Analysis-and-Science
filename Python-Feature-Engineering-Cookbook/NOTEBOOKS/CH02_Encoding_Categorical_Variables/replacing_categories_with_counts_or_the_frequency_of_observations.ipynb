在“用计数或观察频率计数”或频率编码中，我们用计数或显示该类别的观察分数替换类别。也就是说，如果 100 个观察中有 10 个显示颜色变量的蓝色类别，我们在进行计数编码时将蓝色替换为 10，如果进行频率编码则替换为 0.1。**当类别频率和目标之间存在关系时，这些编码方法很有用**。例如，在销售中，产品的频率可能表明其受欢迎程度。

> 如果在相同数量的观测中存在两个不同的类别，它们将被相同的值替换，这可能会导致信息丢失。

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from feature_engine.encoding import CountFrequencyEncoder

data = pd.read_csv("credit_approval_uci.csv")
X_train, X_test, y_train, y_test = train_test_split(
    data.drop(labels=["target"], axis=1),
    data["target"],
    test_size=0.3,
    random_state=0,
)
```

```python
counts = X_train['A7'].value_counts().to_dict()
counts
```

```python
X_train_enc = X_train.copy()
X_test_enc = X_test.copy()

X_train_enc['A7'] = X_train_enc['A7'].map(counts)
X_test_enc['A7'] = X_test_enc['A7'].map(counts)
```

```python
count_enc = CountFrequencyEncoder(encoding_method='count', variables=None, 
                                  missing_values='ignore',
                                  unseen='ignore')
```

> 如果测试集中有训练集中没有的类别，编码器将默认报错。可以忽略它们，在这种情况下它们将显示为 nan，或者将它们编码为 0。

`CountFrequencyEncoder()` will automatically find and encode all categorical variables in the train set. To encode only a subset of the variables, pass the variable names in a list to the `variables` argument. To encode with the frequency instead, use `encoding_method="frequency"`.

```python
count_enc.fit(X_train)
```

```python
count_enc.variables_
```

```python
count_enc.encoder_dict_
```

```python
X_train_enc = count_enc.transform(X_train)
X_test_enc = count_enc.transform(X_test)
```

> 编码的观察计数应从训练集中获取，以避免数据泄露。请注意，测试集中的新类别将没有相应的映射，因此将被替换为 `nan`。为避免这种情况，请使用 feature‑engine。或者，可以将 `nan` 替换为 0。
