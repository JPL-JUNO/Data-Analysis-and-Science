```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('../../DATA/air_passengers.csv', 
                 parse_dates=['ds'], index_col=['ds'])
```

```python
df_imputed = df.interpolate(method='linear')
```

如果行之间的时间间隔不均匀，则应将 `method` 设置为 `time` 以实现线性拟合。

```python
ax = df_imputed.plot(
    linestyle='-', marker='.', figsize=(10, 5)
)
df_imputed[df.isnull()].plot(
    ax=ax, legend=None, marker='.', color='r'
)
ax.set_title("Air passengers")
ax.set_ylabel("Number of passengers")
ax.set_xlabel("Time")
```

```python
df_imputed = df.interpolate(method='spline', order=2)
```

```python
ax = df_imputed.plot(
    linestyle='-', marker='.', figsize=(10, 5)
)
df_imputed[df.isnull()].plot(
    ax=ax, legend=None, marker='.', color='r'
)
ax.set_title("Air passengers")
ax.set_ylabel("Number of passengers")
ax.set_xlabel("Time")
```

> 尽管与前向填充和后向填充相比，插值旨在对缺失数据进行更好的估计，但如果时间序列显示出强烈的趋势和季节性，这些估计可能仍然不准确。
