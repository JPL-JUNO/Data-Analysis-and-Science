```python
import pandas as pd
import random
```

```python
random.seed(123)

list1 = [['A']*3, ['B']*5, ['C']*7]
charlist = [x for sublist in list1 for x in sublist]
random.shuffle(charlist)
ser1 = pd.Series(charlist)
ser1
```

```python
random.seed(123)
ser2 = pd.Series(random.sample(range(10, 100), 15))
ser2
```

```python
prod_df1 = pd.concat([ser1, ser2], axis='columns')
prod_df1.columns = ['Product', 'Sales']
prod_df1.head(5)
```

```python
random.seed(321)
list1 = [['A']*2, ['B']*8, ['C']*5]
charlist = [x for sublist in list1 for x in sublist]
random.shuffle(charlist)
ser3 = pd.Series(charlist)
```

```python
random.seed(321)
ser4 = pd.Series(random.sample(range(10, 100), 15))
ser4.head()
```

```python
prod_df2 = pd.concat([ser3, ser4], axis='columns')
prod_df2.columns = ['Product', 'Sales']
prod_df2.head(5)
```

```python
prod_df2['Sales'].eq(45)
```

```python
prod_df2[prod_df2['Sales'].eq(45)]
```

```python
prod_df2['Sales'].gt(prod_df1['Sales'])
```

```python
prod_df2[prod_df2['Sales'].gt(prod_df1['Sales'])]
```

```python
prod_df2[prod_df2['Sales'].lt(prod_df1['Sales'])]
```

```python
prod_df2.iloc[[2, 5, 6, 8]]
```

```python
tab1 = prod_df1.groupby(['Product']).agg('sum')
tab2 = prod_df2.groupby(['Product']).agg('sum')
```

```python
tab1
```

```python
tab2
```

```python
tab3 = pd.merge(tab1, tab2, on=['Product'])
tab3
```

```python
tab3.columns = ['Sales1', 'Sales2']
tab3
```
