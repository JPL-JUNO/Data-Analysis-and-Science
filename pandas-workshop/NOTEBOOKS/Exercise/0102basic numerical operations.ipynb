```python
import pandas as pd
import numpy as np
import random
```

```python
np.random.seed(123)
```

```python
ser1 = pd.Series(np.random.normal(3.0, 1, 100))
ser2 = pd.Series(np.random.normal(5.0, 3, 100))
ser3 = pd.Series(np.random.normal(1.0, 0.5, 100))
```

```python
df = pd.concat([ser1, ser2, ser3], axis='columns')
df.columns = ['V1','V2','V3']
df.head()
```

```python
df.sum()
```

```python
df.mean()
```

```python
df.apply(lambda ser: divmod(ser, 3))
```

> `apply` 返回的类型会让人感觉非常奇怪。

```python
df.map(lambda element: divmod(element, 3))
```

```python
list1 = [['20']*10, ['35']*15, ['40']*10, ['10']*25, ['15']*40]

char_list = [x for sublist in list1 for x in sublist]
len(char_list)
```

```python
random.seed(123)
```

```python
random.shuffle(char_list)
```

```python
ser4 = pd.Series(char_list)
ser4
```

```python
ser4 = pd.to_numeric(ser4)
ser4
```

```python
df = pd.concat([df, ser4], axis='columns')
df.rename(columns={0: 'V4'}, inplace=True)
df.sample(2)
```

```python
numpy_array = df.to_numpy()
numpy_array
```
