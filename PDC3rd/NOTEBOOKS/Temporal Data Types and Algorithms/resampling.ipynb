`Group By` 允许根据数据集中的唯一值组合拆分数据，对这些分割应用应用算法，然后将结果重新组合在一起。

重新采样与 Group By 非常相似，唯一的区别发生在拆分阶段。重新采样可以获取日期时间并将其分组为每 5 秒或每 10 分钟等增量，而不是通过唯一值组合来生成组。

```python
import pandas as pd
import numpy as np
```

```python
index = pd.date_range(start='2024-01-01', periods=10, freq='s')
ser = pd.Series(range(10), index=index, dtype=pd.Int64Dtype())
ser
```

如果每秒查看这些数据太过细致，可以使用 `pd.Series.resample` 将数据下采样为不同的增量，例如每 3 秒。重新采样还需要使用聚合函数来指示每个增量内的所有记录会发生什么：

```python
ser.resample('3s').sum()
```

对于每个间隔，统计的值为左闭右开。
