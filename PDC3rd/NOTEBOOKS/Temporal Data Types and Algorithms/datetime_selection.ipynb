```python
import pandas as pd
import numpy as np
```

pandas 允许通过与关联行 `pd.Index` 交互从 `pd.Series` 或 `pd.DataFrame` 中选择数据的许多强大方法。如果碰巧使用日期时间数据创建了 `pd.Index`，它最终会表示为一个名为 `pd.DatetimeIndex` 的特殊子类。此子类覆盖了 `pd.Index.loc` 的一些功能方法，提供更灵活的适合时间数据的选择选项。

`pd.date_range` 是一个方便的函数，可快速生成 `pd.DatetimeIndex`。使用此函数的方法之一是使用 `start=` 参数指定开始日期，使用 `freq=` 参数指定步进频率，并使用 `periods=` 参数指定 `pd.DatetimeIndex` 的所需长度（次数）。

```python
pd.date_range(start='2023-12-27', freq='10D', periods=5)
```

A frequency string of `"2W"` will generate dates spaced two weeks apart. If the `start=` parameter is a Sunday, the dates will begin from that date exactly; otherwise, the next Sunday begins the sequence:

> 默认从最近的一个周日开始（含当周周日）

```python
pd.date_range(start='2023-12-27', freq='2W', periods=5)
```

可以通过添加后缀来控制用于固定日期的星期几，例如 `‑WED`，将生成周三而不是周日的日期：

```python
pd.date_range(start='2023-12-27', freq='2W-WED', periods=5)
```

`freq='WOM‑3THU'` 将返回每个月的第三个星期四：

```python
pd.date_range(start='2023-12-27', freq='WOM-3THU', periods=5)
```

可以使用参数 `SMS` 生成每月的第一天和第十五天：

```python
pd.date_range(start='2023-12-27', freq='SMS', periods=5)
```

有无数的字段可以用来描述 `freq`，详情参见[这里](https://pandas.pydata.org/pandas-docs/stable/user_guide/timeseries.html#dateoffset-objects)。

`pd.DatetimeIndex` 的每个元素实际上都是一个 `pd.Timestamp`。当使用它从 `pd.Series` 或 `pd.DataFrame` 中进行选择时，用户最初可能会想写类似下面的代码来选择截至 2024‑01‑18 等日期的所有记录：

```python
index = pd.date_range(start='2023-12-27', freq='10D', periods=20)
ser = pd.Series(range(20), index=index)

ser.loc[:pd.Timestamp('2024-01-18')]
```

类似地，用户可能会倾向于写以下内容来选择日期范围：

```python
ser.loc[pd.Timestamp('2024-01-06'): pd.Timestamp('2024-01-18')]
```

但是，这些从 `pd.DatetimeIndex` 中选择的方法相当冗长。为了方便起见，pandas 允许您传入字符串来表示所需的日期，而不是 `pd.Timestamp` 实例：

```python
ser.loc['2024-01-06': '2024-01-18']
```

也不需要以 `YYYY‑MM‑DD` 格式指定整个日期。例如，如果想要选择 2024 年 2 月的所有日期，只需将字符串 `2024‑02` 传递给 `pd.Series.loc` 调用：

```python
ser.loc['2024-02']
```

```python
ser.loc['2024-02': '2024-03']
```

```python
ser.loc['2024']
```

`pd.DatetimeIndex` 还可以通过提供 `tz=` 参数与时区关联：

```python
index = pd.date_range(start='2024-12-27', freq='12h', periods=6, tz='US/Eastern')
ser = pd.Series(range(6), index=index)

ser
```

当使用字符串从时区敏感的 `pd.DatetimeIndex` 中进行选择时，请注意 pandas 会将字符串参数隐式转换为 `pd.DatetimeIndex` 的时区。

> 字符串表示的是 UTC 时间，要转化一步，这一步是隐式的。

```python
ser.loc[:'2024-12-27 11:59:59']
```

```python
ser.loc[:'2024-12-27 12:00:00']
```
