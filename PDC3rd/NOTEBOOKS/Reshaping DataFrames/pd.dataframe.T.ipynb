转置的主要用例之一是在 apply function 之前将 pd.DataFrame 设置为最佳格式。

```python
df.sum(axis='columns')
```

使用 `axis=1` 参数会大大降低应用程序的性能。如果发现在整个代码中分散了大量 `axis=1` 调用，那么最好先转置数据，然后应用默认 `axis=0` 的函数。

```python
import numpy as np
```

```python
np.random.seed(42)

df = pd.DataFrame(
    np.random.randint(10, size=(2, 10_000)),
    index=list('ab')
)

df
```

```python
df.sum(axis='columns')
```

```python
df.T.sum()
```

如果使用 `axis=1` 作为参数反复进行调用，会发现先转置可以节省大量时间。

```python
import timeit

def baseline_sum():
    for _ in range(100):
        df.sum(axis='columns')
    
    
timeit.timeit(baseline_sum, number=100)
```

```python
def transposed_sum():
    transposed = df.T
    for _ in range(100):
        transposed.sum()
        
    
timeit.timeit(transposed_sum, number=100)
```

总的来说，使用 `pd.DataFrame.T` 来避免后续使用 `axis=1` 的调用是一种非常值得提倡的做法。
